```python
import matplotlib.pyplot as plt
from sklearn.datasets import load_iris
from sklearn.cluster import KMeans

# Manual standard scaling: (X - mean) / std
iris = load_iris()
data = iris.data
X = (data - data.mean(axis=0)) / data.std(axis=0)

m = KMeans(n_clusters=3, random_state=42, n_init=10)
p = m.fit_predict(X)
print("Clusters:", p)
print("Centroids:", m.cluster_centers_)

plt.scatter(X[:, 0], X[:, 1], c=p)
plt.scatter(m.cluster_centers_[:, 0], m.cluster_centers_[:, 1], c='red', marker='X', s=150)
plt.show()
```
